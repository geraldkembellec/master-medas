On va utiliser ici l'OpenGraph Protocol (cf. doc OGP : https://ogp.me/ ).
![rfsp.jpeg](https://ogp.me/logo.png)
Basic Metadata

Pour transformer vos pages Web en objets de type graphe, vous devez ajouter des métadonnées de base à votre page. Nous avons basé la version initiale du protocole OpenGraph sur RDFa, ce qui signifie que vous placerez des balises <meta> supplémentaires dans le <head> de votre page Web. Les quatre propriétés requises pour chaque page sont :

    og:title - The title of your object as it should appear within the graph, e.g., "The Rock".
    og:type - The type of your object, e.g., "video.movie". Depending on the type you specify, other properties may also be required.
    og:image - An image URL which should represent your object within the graph.
    og:url - The canonical URL of your object that will be used as its permanent ID in the graph, e.g., "https://www.imdb.com/title/tt0117500/".

As an example, the following is the Open Graph protocol markup for The Rock on IMDB:

    <html prefix="og: https://ogp.me/ns#">
    <head>
      <title>The Rock (1996)</title>
      <meta property="og:title" content="The Rock" />
      <meta property="og:type" content="video.movie" />
      <meta property="og:url" content="https://www.imdb.com/title/tt0117500/" />
      <meta property="og:image" content="https://ia.media-imdb.com/images/rock.jpg" />
    </head>
    <body>
      ...
    </body>
    </html>

```python
def get_meta_descriptions(urls):
    descriptions = {}
    i=0;

    for url in urls:
        try:
            info={}
            i=i+1
            response = requests.get(url, timeout=10)
            response.raise_for_status()
            soup = BeautifulSoup(response.content, 'html.parser')
            # Balise <meta> dans le <head> on cherche les attributs 'name' et 'description'
            meta_description = soup.find('meta', attrs={'name': 'description'})
            meta_keywords = soup.find('meta', attrs={'name': 'keywords'})
            #print(meta_keywords)
            # On va maintenant utiliser le protocole OpenGraph OGP
            # pareil avec 'og:description'
            # comme valeur de l'attribut 'property'
            # et on récupère aussi la valeur de 'description
            # meta_description_og = soup.find('meta', property={'og:description': 'description'})
            # pareil avec 'og:site_name' comme valeur de l'attribut 'property'
            # meta_og_site_name = soup.find('meta', property={'og:site_name': 'description'})
            # pareil avec 'og:image' comme valeur de l'attribut 'property'
            # meta_og_image = soup.find('meta', property={'og:image': 'description'})
            if meta_description and 'content' in meta_description.attrs:
                descriptions[url] = meta_description['content']
                print('Debug desc '+descriptions[url])
            else:
                descriptions[url] = 'No description found'
                print('Debug desc '+descriptions[url])
            if meta_keywords and 'content' in meta_keywords.attrs:
                print('Debug keywords '+meta_keywords['content'])
            else:
                print('Debug keywords '+'No keyword found')
        except requests.RequestException as e:
            descriptions[url] = f'Error: {e}'
    return descriptions
```

```python
urls=['https://www.doctolib.fr/sessions/new?context=navigation_bar','https://www.semrush.com/','https://www.fdj.fr/', 'https://mabanque.bnpparibas/','https://commission.europa.eu/index_fr']
descriptions = get_meta_descriptions(urls)
for url, description in descriptions.items():
    print(f'{url}: {description}')
#print(descriptions)
```
